## 1. My Lane as an ML Task (Type)

- **Task Type:** Binary Classification & Scoring
- **Definition:** Predict whether a given Shopify store landing page or product page is at high risk of low checkout conversion due to page performance, layout friction, or SEO content gaps.
- **Goal:** Automatically flag underperforming store pages for technical audits and optimization.

## 2. Target or Proxy

- **Target Variable (`is_high_risk_conversion`):**
  - `1` = High Risk / Low Converting Page (e.g., Page Load Speed > 3.5s, High Bounce Rate, Missing Meta/Product Description, Low Mobile Usability Score).
  - `0` = Optimized / High Converting Page.
- **Proxy Rationale:** Real-time conversion rate data is often restricted or delayed. Therefore, technical performance metrics (LCP, CLS, DOM depth, word count, schema tags) act as a strong proxy for user friction and conversion failure.

## 3. Success Metric

- **Primary Metric:** **Precision @ Top K (and Recall for High-Risk Pages)**
- **Business Rationale:** False Positives waste engineering time auditing store pages that are already fine. We need high Precision so that when the system flags a page as "High Risk," it almost certainly needs technical refactoring.
- **Secondary Metric:** F1-Score to maintain balance between catching broken pages and avoiding false alarms.

## 4. The Unit of Analysis (Real Dataframe)

- **Unit of Analysis:** **ONE ROW = ONE SHOPIFY STORE URL / PAGE**

```python
import pandas as pd
import numpy as np

# Starter data slice for FlyRank / Store Audit Lane
data = {
    'page_id': ['PAGE_101', 'PAGE_102', 'PAGE_103', 'PAGE_104', 'PAGE_105'],
    'page_type': ['product', 'landing', 'collection', 'product', 'landing'],
    'load_time_seconds': [4.2, 1.8, 2.5, 5.1, 1.2],
    'mobile_usability_score': [45, 92, 78, 38, 95],
    'word_count': [120, 850, 310, 80, 1100],
    'has_schema_markup': [0, 1, 1, 0, 1],
    'bounce_rate_pct': [68.5, 24.1, 41.2, 79.0, 18.5]
}

df = pd.DataFrame(data)

print("--- Section 4: Unit of Analysis (Real Dataframe) ---")
print("Unit of Analysis: One row represents one unique Shopify Store URL / Page.")
df
```

```python
# Target Proxy Sketching
# High Risk (1) if Load Time > 3.5s OR Mobile Score < 50
df['target_high_risk'] = np.where((df['load_time_seconds'] > 3.5) | (df['mobile_usability_score'] < 50), 1, 0)

print("--- Dataframe with Target Proxy Column (`target_high_risk`) ---")
df[['page_id', 'page_type', 'load_time_seconds', 'mobile_usability_score', 'target_high_risk']]
```

## 5. Why ML Beats a Fixed Rule Here

1. **Non-Linear Interacting Factors:** A static rule like `if load_time > 3s THEN alert` fails because a page with a 3.2s load time might still convert extremely well if it has exceptional social proof, clear schema, and high mobile usability. ML captures complex non-linear feature interactions.
2. **Adaptability Across Niches:** Different e-commerce niches (e.g., fashion vs. high-ticket electronics) have different user tolerances for page layout and speed. ML models adapt to distribution shifts faster than hardcoded `if-else` scripts.
3. **Action Triggered:** The ML prediction directly triggers an automated **Audit & Refactoring Brief** sent to the client or developer, prioritizing pages with the highest predicted risk score.
